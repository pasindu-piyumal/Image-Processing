# Image

Image is a set of pixels.
Basic image types:

    Color images (Ex-B,G,R(0-255)) There are 3 components. There are 24 bits(B-8bit, ..)

    Gray images (0-255). Include one 8 bit number

    Binary images (0 or 1)

## 1. Opencv Basics

### How to load a image

```python
import cv2 #import opencv module

#absolute path of the image
img = cv2.imread('../Samples/flower.jpg') #read image

# print(img.shape) #print image array
print(img[100,100]) #print pixel value at row 100 and column 100(B G R format)
# img[100,100] = [255,255,255] #change pixel value at row 100 and column 100 to white
img[100:120, 70:150] = [255,255,255] #change pixel value at row 100 to 120 and column 70 to 150 to white

cv2.imshow('Live', img)
cv2.waitKey(0) #wait for a key press to exit
# cv2.waitkey(5) close the window after 5 milliseconds
cv2.destroyAllWindows() #close all windows
```

###  Writing and Saving Images
Use function cv2.inwrite() to save an image

First argument is the file, second argument is the image you want to save

```python
img = cv2.imread('../Samples/flower.jpg')
img[100:120, 70:150] = [0,255,255] #change pixel value at row 100 to 120 and column 70 to 150 to yellow
cv2.imshow('new-flower-2.png', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Use ascii values for key stop video

```python
import cv2
source = cv2.VideoCapture('../Samples/facevideo.mp4')
while(True):
    ret, img = source.read()
    if (ret==False):
        break
    cv2.imshow('IMG', img)
    key = cv2.waitKey(1)

    if (key==27):
        break
cv2.destroyAllWindows()
```

Capture an image while playing video

```python
import cv2
source = cv2.VideoCapture('../Samples/facevideo.mp4')
while(True):
    ret, img = source.read()
    if (ret==False):
        break
    cv2.imshow('IMG', img)
    key = cv2.waitKey(1)

    if (key==27):
        break

    if (key==112):
        cv2.imwrite('capture.jpg', img)
cv2.destroyAllWindows()
```

# Image Processing
## 1 - Point Operation

### Color Conversion

```python
img1 = cv2.imread('../Samples/flower.jpg')
img2 = cv2.imread('../Samples/flower.jpg', 0)

print(img1.shape)
print(img2.shape)

cv2.imshow('IMG1', img1)
cv2.imshow('IMG2', img2)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Changing Colorspace

```python
img = cv2.imread('../Samples/flower.jpg')
img2 = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

cv2.imshow('IMG', img)
cv2.imshow('IMG2', img2)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread('../Samples/flower.jpg')
img2 = cv2.cvtColor(img, cv2.COLOR_RGB2HSV)

cv2.imshow('IMG', img)
cv2.imshow('IMG2', img2)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
